### **Question 15**

```python
df["Runtime"] = df["Runtime"].replace("min", "").astype(int)

df["Runtime"] # .str is not used due to already string converted into integer
```

### **Question 16**
'Gross' column mein jo missing values (NaN) hain, unko pooray column ke median value se fill (impute) karein.

```python
df["Gross"] = pd.to_numeric(df["Gross"].astype(str).str.replace(",", ""), errors="coerce")

df["Gross"] = df["Gross"].fillna(df["Gross"].median())

df["Gross"]
```

### **Question 21**
(Visualization) IMDB_Rating ka distribution dekhne ke liye ek neat Histogram plot karein.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.hist(df["IMDB_Rating"], bins=15)
plt.title("Distribution of IMDB Ratings")
plt.xlabel("IMDB Rating")
plt.ylabel("Frequency")
plt.show()
```

### **Question 22**
(Visualization) Top 10 Directors aur unki total movies ka ek attractive Bar Chart banayein.

```python
import matplotlib.pyplot as plt

top_directors = df["Director"].value_counts().head(10)

plt.figure(figsize=(10,5))
top_directors.plot(kind="bar")
plt.title("Top 10 Directors")
plt.xlabel("Director")
plt.ylabel("Number of Movies")
plt.xticks(rotation=45)
plt.show()
```

### **Question 23**
(Visualization) IMDB_Rating aur Gross revenue ke darmiyan ek Scatter Plot banayein taake dono ka relation dikhe.

```python
import matplotlib.pyplot as plt

df["Gross"] = pd.to_numeric(df["Gross"].astype(str).str.replace(",", ""), errors="coerce")

plt.figure(figsize=(8,5))
plt.scatter(df["IMDB_Rating"], df["Gross"])
plt.title("IMDB Rating vs Gross")
plt.xlabel("IMDB Rating")
plt.ylabel("Gross")
plt.show()
```

### **Question 24**
(Visualization) Genres ka count dekhne ke liye top 5 genres ka ek clean Pie Chart banayein.

```python
import matplotlib.pyplot as plt

genres = df["Genre"].str.split(", ").explode().value_counts().head(5)

plt.figure(figsize=(6,6))
plt.pie(genres, labels=genres.index, autopct="%1.1f%%", startangle=90)
plt.title("Top 5 Genres")
plt.show()
```

### **Question 25**
(Visualization) Meta_score ka distribution aur outliers check karne ke liye ek Box Plot banayein.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))
plt.boxplot(df["Meta_score"].dropna())
plt.title("Meta Score Box Plot")
plt.ylabel("Meta Score")
plt.show()
```

### **Question 26**
(Visualization) Dataset mein 'Certificate' wise average 'Gross' kamai ka ek Bar Plot banayein.

```python
import matplotlib.pyplot as plt

avg_gross = df.groupby("Certificate")["Gross"].mean().sort_values(ascending=False)

plt.figure(figsize=(8,5))
avg_gross.plot(kind="bar")
plt.title("Average Gross by Certificate")
plt.xlabel("Certificate")
plt.ylabel("Average Gross")
plt.xticks(rotation=45)
plt.show()
```

### **Question 27**
(Visualization) Har saal (Released_Year) kitni movies release hueen, iska ek Line Plot banayein.

```python
import matplotlib.pyplot as plt

movies_per_year = df["Released_Year"].value_counts().sort_index()

plt.figure(figsize=(10,5))
plt.plot(movies_per_year.index, movies_per_year.values)
plt.title("Movies Released Per Year")
plt.xlabel("Year")
plt.ylabel("Number of Movies")
plt.show()
```

### **Question 28**
(Visualization) 'No_of_Votes' ke mutabiq top 5 highest voted movies ka ek Horizontal Bar Chart banayein.

```python
import matplotlib.pyplot as plt

top_votes = df.nlargest(5, "No_of_Votes")

plt.figure(figsize=(8,5))
plt.barh(top_votes["Series_Title"], top_votes["No_of_Votes"])
plt.title("Top 5 Highest Voted Movies")
plt.xlabel("Votes")
plt.ylabel("Movie")
plt.show()
```
